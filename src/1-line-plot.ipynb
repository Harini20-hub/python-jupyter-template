### Exercise 1: Line Plot Customization

```python
import matplotlib.pyplot as plt
# Data
x = list(range(11))
y = [i**2 for i in x]

# Create the plot
plt.plot(x, y)

# Customize the line plot
plt.title('Square Numbers Line Plot')
plt.xlabel('X values')
plt.ylabel('Y = X Squared')
plt.grid(True)

# Show the plot
plt.show()
```
